```python
# 1. Que categorias tienen mas ventas, aca se empieza a utilizar merge para unir tablas
df3.head()
```

```python
df7.head()
```

```python
categorias_que_mas_vendieron = df7.merge(df3, on="product_id")
```

```python
#Creo una columna que integre el valor totoal. Le sumo el costo de envio que es el freight value
categorias_que_mas_vendieron["Valor_total"] = categorias_que_mas_vendieron["price"] + categorias_que_mas_vendieron["freight_value"]
```

```python
categorias_que_mas_vendieron.groupby("product_category_name")["Valor_total"].sum().sort_values(ascending=False).head(10)
```

```python
# 2. Clientes que mas gastan
tabla_clientes = df.merge(df6, on="customer_id")
```

```python
pago_total = tabla_clientes.merge(df4, on="order_id")
```

```python
pago_total.groupby("customer_id")["payment_value"].sum().sort_values(ascending=False).head(10)
```

```python
# 3. Evolucion de ventas por mes
df6["order_purchase_timestamp"] = pd.to_datetime(df6["order_purchase_timestamp"])
```

```python
df6["month"] = df6["order_purchase_timestamp"].dt.month
```

```python
df_evolucion_mes = df6.merge(df4, on="order_id")
```

```python
df_evolucion_mes.groupby("month")["payment_value"].sum()
```

```python
# 4. metodos de pago mas utilizados
df_pagos_reales = df4.merge(df6[['order_id', 'order_status']], on='order_id')
```

```python
df_entregados = df_pagos_reales[df_pagos_reales['order_status'] == 'delivered']
```

```python
df_entregados['payment_type'].value_counts()
```

```python
# 5. Tiempo promedio de entrega
df6['order_purchase_timestamp'] = pd.to_datetime(df6['order_purchase_timestamp'])
df6['order_delivered_customer_date'] = pd.to_datetime(df6['order_delivered_customer_date'])
```

```python
df6['tiempo_entrega_real'] = df6['order_delivered_customer_date'] - df6['order_purchase_timestamp']
```

```python
df6['tiempo_entrega_real'].dt.days.mean()
```

```python
# 6. Estados o ciudades con mas ventas
estados_mas_ventas = df.merge(df6, on="customer_id").merge(df4, on="order_id")
estados_mas_ventas = estados_mas_ventas[estados_mas_ventas['order_status'] == 'delivered']
estados_mas_ventas.groupby('customer_city')['payment_value'].sum().sort_values(ascending=False).head(10)
```

```python
#Graficos
#1 Categorias que mas vendieron
top_categorias = (categorias_que_mas_vendieron
                  .groupby("product_category_name")["Valor_total"]
                  .sum()
                  .sort_values(ascending=False)
                  .head(10))

# 2. Graficamos la nueva variable 'top_categorias'
top_categorias.plot(kind="bar", figsize=(10, 5), color="skyblue", edgecolor="black")

# 3. Configuraciones de títulos y etiquetas
plt.title("Top 10 Categorías con más Ventas", fontsize=14, fontweight='bold')
plt.xlabel("Categorías", fontsize=11)
plt.ylabel("Valor Total", fontsize=11)
plt.xticks(rotation=45, ha="right") # se rotan ligeramente los nombre para que se lean bien

# 4. Guardamos y mostramos de forma limpia
plt.tight_layout()
plt.savefig("images/categorias_mas_vendieron.png")
plt.show()
```

```python
# .2 Clientes que mas gastan
# Se que queda muy feo mostrar los id de los clientes pero no tenian otro dato que los destaque
pago = pago_total.groupby("customer_id")["payment_value"].sum().sort_values(ascending=False).head(10)

pago.plot(kind="bar", figsize=(10,5), color="skyblue", edgecolor="black")

plt.title("Top 10 Clientes que más Gastan", fontsize=14, fontweight='bold')
plt.xlabel("Clientes", fontsize=11)
plt.ylabel("Total", fontsize=11)
plt.xticks(rotation=45, ha="right") # aca igual que en la primer consulta


plt.tight_layout()
plt.savefig("images/clientes_mas_gastan.png")
plt.show()
```

```python
# 3 Evolucion de ventas por mes
evolucion_mes = df_evolucion_mes.groupby("month")["payment_value"].sum().head(12)

evolucion_mes.plot(kind="line", figsize=(10,5), color="skyblue")

plt.title("Evolucion de ventas por mes", fontsize=14, fontweight='bold')
plt.xlabel("Meses", fontsize=11)
plt.ylabel("Ventas", fontsize=11)

plt.xticks(range(1, 13), rotation=45, ha="right") 

plt.tight_layout()
plt.savefig("images/Evolucion_ventas_por_mes.png")
plt.show()
```

```python
metodos_pago = df_entregados["payment_type"].value_counts()
metodos_pago = (
    df_entregados
    .groupby("payment_type")
    .size()
    .sort_values(ascending=False)
)
metodos_pago.plot(
    kind="bar",
    figsize=(10,5),
    color="skyblue",
    edgecolor="black"
)
plt.title("Metodos de pago mas utilizados", fontsize=14, fontweight='bold')
plt.xlabel("Metodos", fontsize=11)
plt.ylabel("Cantidad", fontsize=11)

plt.xticks(rotation=45, ha="right")

plt.tight_layout()

plt.savefig("images/metodos_de_pago_mas_utilizados.png")

plt.show()
```
