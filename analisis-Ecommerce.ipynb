```python
# 1. Que categorias tienen mas ventas, aca se empieza a utilizar merge para unir tablas
df3.head()
```

```python
df7.head()
```

```python
categorias_que_mas_vendieron = df7.merge(df3, on="product_id")
```

```python
#Creo una columna que integre el valor totoal. Le sumo el costo de envio que es el freight value
categorias_que_mas_vendieron["Valor_total"] = categorias_que_mas_vendieron["price"] + categorias_que_mas_vendieron["freight_value"]
```

```python
categorias_que_mas_vendieron.groupby("product_category_name")["Valor_total"].sum().sort_values(ascending=False)
```

```python
# 2. Clientes que mas gastan
tabla_clientes = df.merge(df6, on="customer_id")
```

```python
pago_total = tabla_clientes.merge(df4, on="order_id")
```

```python
pago_total.groupby("customer_id")["payment_value"].sum().sort_values(ascending=False) 
```

```python
# 3. Evolucion de ventas por mes
df6["order_purchase_timestamp"] = pd.to_datetime(df6["order_purchase_timestamp"])
```

```python
df6["month"] = df6["order_purchase_timestamp"].dt.month
```

```python
df_evolucion_mes = df6.merge(df4, on="order_id")
```

```python
df_evolucion_mes.groupby("month")["payment_value"].sum()
```

```python
# 4. metodos de pago mas utilizados
df_pagos_reales = df4.merge(df6[['order_id', 'order_status']], on='order_id')
```

```python
df_entregados = df_pagos_reales[df_pagos_reales['order_status'] == 'delivered']
```

```python
df_entregados['payment_type'].value_counts()
```

```python
# 5. Tiempo promedio de entrega
df6['order_purchase_timestamp'] = pd.to_datetime(df6['order_purchase_timestamp'])
df6['order_delivered_customer_date'] = pd.to_datetime(df6['order_delivered_customer_date'])
```

```python
df6['tiempo_entrega_real'] = df6['order_delivered_customer_date'] - df6['order_purchase_timestamp']
```

```python
df6['tiempo_entrega_real'].dt.days.mean()
```

```python
# 6. Estados o ciudades con mas ventas
estados_mas_ventas = df.merge(df6, on="customer_id").merge(df4, on="order_id")
estados_mas_ventas = estados_mas_ventas[estados_mas_ventas['order_status'] == 'delivered']
estados_mas_ventas.groupby('customer_city')['payment_value'].sum().sort_values(ascending=False)
```

```python
#Graficos
#1 Categorias que mas vendieron
categorias_que_mas_vendieron.plot(kind="bar")
plt.title("Categorias que mas vendieron")
plt.xlabel("Categorias")
plt.ylabel("Cantidad")
plt.savefig("images/categorias_mas_vendieron.png")
plt.show()
```
